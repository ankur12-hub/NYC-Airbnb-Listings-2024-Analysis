### Task 4: Data Cleaning

```python
# dealing with missing values it is showing zero because we have used (inplace = True) in below code
data.isnull().sum()
```

```python
# Now after dropping th missing values entire rows we are again using this code that we used above
data.isnull().sum()
```

```python
# identifying outliers in price
sns.boxplot(data=data, x='price')
# here we can see there is an outlier which has a price 100000 . Most of the price lies between 0 to 20000 but one is at 100000
```

```python
df.dtypes
```

```python
# heat map - correlation of one variable with other for numerical column
#if the correlation is negative means if one parameter increases other decreases and if its positive means if one parameter is increases other also increase.
corr = df[['latitude','longitude','price','minimum_nights','number_of_reviews','reviews_per_month','availability_365','beds']].corr()
corr
```

```python
sns.heatmap(data=corr,annot=True)
plt.figure(figsize=(8,6))
# if the correlation is 1 it is perfectly positive correlation if the correlation is -1 it is perfectly negative correlation         if it is 0 means there is no relationship between 2 columns. lets see price with number_of_reviews  or price with beds its is highlighted and also having positive value means if the price increases the number of beds also increases and vice-versa . If you see price with number of reviews you can see that it is -0.059 means if price increases the number of reviews value decreases
```
